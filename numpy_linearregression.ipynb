```
import numpy as np
import matplotlib.pyplot as plt
import imageio.v2 as imageio
import os
```

```
data_x = np.linspace(-10.0, 10.0, 100)[:, np.newaxis]
# data_y = np.sin(data_x) + 0.1 * np.power(data_x, 2) + 0.5 * np.random.randn(100, 1)
data_y = 40*data_x + 20* np.random.randn(100, 1)  # creating your data
data_x /= np.max(data_x) #scaling
data_x
```

```
from scipy import stats
a = np.array([0,1,2,3,4,5,6, 23, 19, 11])
b = np.array([0, 1, 4, 9, 16, 25, 36, 144, 225, 196])
stats.pearsonr(a,b)
```

```
data_x = np.hstack((np.ones_like(data_x), data_x))
data_x
```

```
order = np.random.permutation(len(data_x))
portion = 20
test_x = data_x[order[:portion]]
test_y = data_y[order[:portion]]
train_x = data_x[order[portion:]]
train_y = data_y[order[portion:]]
print(order)
print (test_x.shape, train_x.size, test_y.size, train_y.size)
print (test_x)
```

```
plt.scatter(train_x[:,1], train_y, c='b', label='Data')
plt.show()
```

```
def get_gradient(w, x, y):
    y_estimate = (np.power(x,1).dot(w)).flatten() #hypothesis
    error = (y.flatten() - y_estimate)
    mse = (1.0/len(x))*np.sum(np.power(error,2)) # mse
    gradient = -(1.0/len(x)) * error.dot(np.power(x,1))         # gradient
    return gradient, mse
```

```
w = np.random.randn(2)  # Random Initialization
alpha = 0.4  # learning rate
tolerance = 1e-2  # param for stopping the loop
print("Intial values of Weights:")
print(w[1], w[0])

# Create a directory to save the plots
if not os.path.exists('training_plots'):
    os.makedirs('training_plots')

# Perform Gradient Descent
iterations = 1
filenames = []
while True:
    gradient, error = get_gradient(w, train_x, train_y)
    new_w = w - alpha * gradient

    # Save plot at each iteration
    plt.figure()
    plt.plot(data_x[:, 1], (data_x**1).dot(new_w), c='g', label='Model')
    plt.scatter(train_x[:, 1], train_y, c='b', label='Train Set')
    plt.grid()
    plt.legend(loc='best')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title(f'Iteration {iterations}: W1={new_w[1]:.4f}, W0={new_w[0]:.4f}') # Add title with weights and iteration
    filename = f'training_plots/iteration_{iterations:04d}.png'
    plt.savefig(filename)
    plt.close()
    filenames.append(filename)

    # Stopping Condition
    if np.sum(abs(new_w - w)) < tolerance:
        print("Converged at iteration: ", iterations)
        break

    # Print error every 10 iterations
    if iterations % 10 == 0:
        print("Iteration: %d - Error: %.4f" % (iterations, error))
        print("Updated Weights : {:f} , {:f}".format(new_w[1], new_w[0]))

    # Print weights at each iteration
    print(f"Iteration {iterations}: W1={new_w[1]:.4f}, W0={new_w[0]:.4f}")

    iterations += 1
    w = new_w


print("Final Weights : {:f} , {:f}".format(w[1], w[0]))
print("Test Cost =", get_gradient(w, test_x, test_y)[1])

# Create GIF from saved plots
with imageio.get_writer('training_progress.gif', mode='I') as writer:
    for filename in filenames:
        image = imageio.imread(filename)
        writer.append_data(image)

# Remove individual plot files
for filename in filenames:
    os.remove(filename)

print("GIF created: training_progress.gif")
```
