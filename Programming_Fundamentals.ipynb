```python
Low Level vs High Level
1. System can understand Easily - LL
1. User can understand Easily - HL

2. Architecture Dependent - LL
2. Not Architecture Dependent - HL

3. Bit Complex to code and hard to manage - LL
3. Easier - HL

4. Security Concerns, Managing the memory - LL
4. These things can be avoided - HL

5. Neither Portable nor Platform independent - LL
5. Are portable and may be platform independent - HL
```
